```python
# TASK 2: Simple Classification Model - Iris Dataset
# CodeOrbit ML Internship - By Gayathri Srirangam

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, classification_report, confusion_matrix

print("--- TASK 2: Simple Classification Model ---")

# 1. Load Dataset (Small public dataset - Iris)
iris = load_iris()
X = pd.DataFrame(iris.data, columns=iris.feature_names)
y = pd.Series(iris.target)
print("Dataset Loaded: Iris Flower Dataset")
print(f"Shape: {X.shape}")
print(f"Target Classes: {iris.target_names}")

# 2. Split data into Training and Testing (70% train, 30% test)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
print(f"\nTraining set: {X_train.shape}, Testing set: {X_test.shape}")

# 3. Build Model - Logistic Regression (as per task instruction)
model = LogisticRegression(max_iter=200)
model.fit(X_train, y_train)
print("\nModel Trained: Logistic Regression")

# 4. Prediction
y_pred = model.predict(X_test)

# 5. Evaluation - Accuracy, Precision, Recall (as per task)
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred, average='weighted')
rec = recall_score(y_test, y_pred, average='weighted')

print("\n--- MODEL EVALUATION RESULTS ---")
print(f"Accuracy : {acc*100:.2f}%")
print(f"Precision: {prec:.4f}")
print(f"Recall   : {rec:.4f}")

print("\nDetailed Classification Report:")
print(classification_report(y_test, y_pred, target_names=iris.target_names))

# 6. Confusion Matrix Visualization
plt.figure(figsize=(6,4))
sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, cmap='Blues',
            xticklabels=iris.target_names, yticklabels=iris.target_names)
plt.title('Confusion Matrix - Iris Classification')
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()

# 7. Brief Explanation of Results
print("""
--- BRIEF EXPLANATION ---
We used Logistic Regression on Iris dataset to classify 3 flower species.
Model achieved high accuracy (>90%), showing it can correctly distinguish
Setosa, Versicolor, Virginica based on petal/sepal measurements.
Precision and Recall are balanced, indicating no major bias.
This satisfies Task 2 requirement: Build, Split, Evaluate.
""")
```
